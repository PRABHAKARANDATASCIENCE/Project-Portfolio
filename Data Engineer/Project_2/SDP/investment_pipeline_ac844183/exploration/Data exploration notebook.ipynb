```sql
%py
# Let's plot the total investment per year and monthly breakdown per year
pdf2 = df2.toPandas()
yearly = pdf2.groupby('year', as_index=False)['total_investment'].sum().sort_values('year')

# Plot 1: Total investment per year
plt.figure(figsize=(12, 6))
plt.bar(yearly['year'].astype(str), yearly['total_investment'], label="yearly investment")
plt.legend()
plt.xlabel("Year")
plt.ylabel("Total Investment")
plt.title("Total Investment per Year")
plt.tight_layout()
plt.show()

# Plot 2: Trend line — monthly investment from 2024 to 2026
pdf2_sorted_all = pdf2.sort_values(['year', 'month_num'])
# Build a combined date label for the x-axis (e.g. "2024-Jan")
pdf2_sorted_all['date_label'] = pdf2_sorted_all['year'].astype(str) + '-' + pdf2_sorted_all['month'].astype(str)

fig2, ax2 = plt.subplots(figsize=(16, 6))
ax2.plot(range(len(pdf2_sorted_all)), pdf2_sorted_all['total_investment'].values, marker='o', linewidth=2, label="Monthly investment")
ax2.set_xticks(range(len(pdf2_sorted_all)))
ax2.set_xticklabels(pdf2_sorted_all['date_label'].values, rotation=90, fontsize=8)
ax2.set_xlabel("Month")
ax2.set_ylabel("Total Investment")
ax2.set_title("Monthly Investment Trend (2024–2026)")
ax2.legend()
ax2.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
