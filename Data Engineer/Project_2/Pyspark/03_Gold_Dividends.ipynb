```python
%sql
select Symbol,count(*) total_dividends,
sum(total_dividend) total_dividend_income,
avg(total_dividend) avg_dividend_income,
min(Ex_date) as first_dividend,
max(Ex_date) as last_dividend
from investment_pyspark.silver.dividens_cleaned
group by Symbol
order by total_dividend_income desc;
```

```python
%sql
create or replace table investment_pyspark.gold.monthly_dividend as (
select date_trunc('year',Ex_date) _year,
date_trunc('month',Ex_date) _month,
sum(Total_dividend) as dividend_income,
count(*) as number_of_dividends
from investment_pyspark.silver.dividens_cleaned
group by date_trunc('year',Ex_date),date_trunc('month',Ex_date))
--order by date_trunc('year',Ex_date);
```
