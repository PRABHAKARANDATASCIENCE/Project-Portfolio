```sql
create schema if not exists investment.silver;
```

```sql
DESCRIBE TABLE investment.bronze.holdings_raw;
```

```sql
select * from investment.bronze.holdings_raw
limit 10;
```

```sql
CREATE OR REPLACE TABLE investment.silver.holdings_clean AS
SELECT 
  UPPER(TRIM(Instrument)) AS instrument,
  CASE 
    WHEN UPPER(TRIM(Instrument)) RLIKE '^[0-9]' 
      OR UPPER(TRIM(Instrument)) RLIKE '(BOND|SGB|SGL|MML|NFL|KFL)' THEN 'DEBT'
    WHEN UPPER(TRIM(Instrument)) RLIKE '(BEES|ETF|MON100)' THEN 'ETF'
    ELSE 'EQUITY' 
  END AS asset_class,
  CAST(Qty AS INT) AS quantity,
  CAST(Avg_cost AS DECIMAL(18, 2)) AS avg_cost,
  CAST(LTP AS DECIMAL(18, 2)) AS current_price,
  CAST(Invested AS DECIMAL(18, 2)) AS invested_value,
  CAST(Cur_val AS DECIMAL(18, 2)) AS current_value,
  CAST(`P&L` AS DECIMAL(18, 2)) AS unrealized_pnl,
  CAST(Net_chg AS DECIMAL(8, 4)) AS net_change_pct,
  CAST(Day_chg AS DECIMAL(8, 4)) AS day_change_pct,
  ingestion_timestamp AS bronze_ingestion_timestamp,
  current_timestamp() AS silver_ingestion_timestamp
FROM investment.bronze.holdings_raw
WHERE Instrument IS NOT NULL 
  AND Qty > 0;
```

```sql
select * from investment.silver.holdings_clean
limit 10;
```
