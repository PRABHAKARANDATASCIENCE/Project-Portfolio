```sql
create table if not exists investment.bronze.holdings_raw as 
select 
  Instrument,
  `Qty.` as Qty,
  `Avg. cost` as Avg_cost,
  LTP,
  Invested,
  `Cur. val` as Cur_val,
  `P&L`,
  `Net chg.` as Net_chg,
  `Day chg.` as Day_chg,
  _rescued_data,
  current_timestamp() as ingestion_timestamp,
  _metadata.file_path as source 
from read_files('/Volumes/investment/bronze/raw_data/holdings.csv',format =>'csv',header=> true,inferSchema => true);
```
