```sql
create schema if not exists investment.gold;
```

```sql
create or replace table investment.gold.holdings as
select asset_class, instrument, quantity,
sum(invested_value) as total_invested,
sum(current_value) as total_current_value,
sum(unrealized_pnl) as total_unlrealized_pnl,
coalesce(round((sum(unrealized_pnl)/nullif(sum(invested_value),0))*100,2), 0) as overall_return_pct
from investment.silver.holdings_clean
group by asset_class, instrument, quantity
order by overall_return_pct desc;
```
